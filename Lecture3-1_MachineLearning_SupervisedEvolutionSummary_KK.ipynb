# 알고리즘 정확성 vs. 설명력 관계

**1) 기계학습 활용 데이터분석의 현실:**

> **"(이상적으로) 머신러닝 알고리즘에 데이터를 `학습/적합/모델링` 한다는 건.."**     
>
> (1) 사람/사물/시스템이 `어떻게 동작하는지 이해`의 과정    
> (2) 사람/사물/시스템이 만들어내는 `데이터를 체계적으로 요약`하는 과정    
> (3) 미래의 예측값과 실제값의 비교로 `사람/사물/시스템을 일반화`하는 과정    
> (4) 일반화된 사람/사물/시스템으로 더욱 `효과적이고 체계적인 의사결정`하는 방법    

> **"(현실적으로) 많은 시간과 비용이 투입되지만, `우리 사회를 이해((1),(2),(3)) <<< 빠른 의사결정(4)`에 집중되어 효과는 글쎄.."**
>
> - 사회를 이해하기 위한 `사회학/교육학/철학/경제학 등`의 학문은 `((1),(2),(3))에 집중`
> - `경영과학/컴퓨터과학/산업공학 등`의 학문은 기술적인 자동화나 인공지능을 반영하는 `(4)에 집중`
> - 학문적 출신(?)에 따라 `분석에 대한 관점 차이` 또는 `비즈니스 방향 차이`가 존재할 수 있음

# **알고리즘 발전 이해를 위한 비용함수 구조**

**0) 지도학습 문제해결 비교:** `회귀분석 vs 분류분석`

- **Comparison Summary:** `데이터분석 단계`는 동일하지만 `비용함수나 검증지표 등`에서 차이

| **구분** | **Linear Regression** | **Logistic Regression** |
|:---:|:---:|:---:|
| **문제 유형<br>(해결 방향)** | 회귀<br>(Regression) | 분류<br>(Classification) |
| **모델 가정<br>(전문가 이상형)** | 직선<br>(Linear) | 곡선<br>(Sigmoid) |
| **비용 함수<br>(평가 기준)** | 정답과 예측의 차이가 최소인 직선 선택<br>(MSE) | 각 라벨을 잘 맞춘 곡선 선택<br>(Entropy) |
| **검증 지표<br>(평가 확대)** | 정답과 예측의 차이가 얼마나 최소인가?<br>(RMSE, MSPE, MAE, MAPE, MedAE, MedAPE 등) | 각 라벨을 얼마나 잘 맞췄나?<br>(Precision, Recall, F1-score, Accuracy, AUC, ROC 등) |

---

**1) 수학적 이해:** `비용함수(Cost Function)` = `편향(Bias)` + `분산(Variance)` + `에러(Error)`

<center><img src='Image/Expert/Bias-Variance_Structure.PNG' width='500'></center>

> - $F$: `정답` 알고리즘, $\hat{F}$: `예측`을 위해 데이터를 `학습`한 알고리즘
> 
> \begin{align*}
\text{Real Y} && Y &= F(x) \\
\text{Estimated Y} && \hat{Y} &= \hat{F}(x) + \epsilon, ~~~ \epsilon \sim N(0, \sigma^2) \\
\end{align*}

\begin{align*}
\text{Cost Function} && MSE &= E\Bigl[\bigl(Y-\hat{Y})^2 \Bigr] \\
&& &= E\Bigl[\bigl(F(x)-\hat{F}(x)-\epsilon\bigr)^2 \Bigr] \\
&& &= E\Bigl[\bigl([F(x) - \hat{F}(x)] - \epsilon\bigr)^2 \Bigr] \\
&& &= E\Bigl[\bigl(F(x) - \hat{F}(x)\bigr)^2 \Bigr] - 2 \bigl(F(x) - \hat{F}(x)\bigr) E(\epsilon) + E(\epsilon^2) \\
&& &= E\Bigl[\bigl(F(x) - \hat{F}(x)\bigr)^2 \Bigr] + \sigma^2 \\
&& &= E\Bigl[\bigl(F(x) - E\bigl[\hat{F}(x)\bigr] + E\bigl[\hat{F}(x)\bigr] - \hat{F}(x)\bigr)^2 \Bigr] + \sigma^2 \\
&& &= E\Bigl[\bigl(F(x) - E\bigl[\hat{F}(x)\bigr] \bigr)^2 \Bigr] + E\Bigl[\bigl(E\bigl[\hat{F}(x)\bigr] - \hat{F}(x)\bigr] \bigr)^2 \Bigr] + \sigma^2 \\
&& &= E\Bigl[\bigl(F(x) - \bar{F}(x) \bigr)^2 \Bigr] + E\Bigl[\bigl(\bar{F}(x) - \hat{F}(x)\bigr] \bigr)^2 \Bigr] + \sigma^2 \\
&& &= \text{Bias}^2 \bigl( \hat{F}(x) \bigr) + \text{Variance} \bigl( \hat{F}(x) \bigr) + \text{Irreducible Natural Error}
\end{align*}

<span style="color:red">$\Rightarrow$ **"`비용함수를 줄이려면` 제거하기 어려운 에러를 뺀 `편향`과 `분산`을 줄이는 것이 최선"**</span>

---

**2) 비수학적 이해:** 직관적으로 `편향 + 분산`이 `모두 작은` 경우 `알고리즘 성능 Best`

> **편향(Bias):** `여러번 측정`된 정확성의 `중심 통계량(점추정)` 
>
> - 여러번 모델링을 통해 데이터를 반복 학습 및 추정했을 때, `예측값이 실제값에서 얼마나 떨어져있는지` 측정
> - `데이터 패턴을 잘 반영`했는지 측정
>
> **분산(Variance):** `여러번 측정`된 정확성의 `변동 통계량(구간추정)`  
>
> - 여러번 모델링을 통해 데이터를 반복 학습 및 추정했을 때, 실제값과 상관없이 `예측값들이 퍼져있는 정도`
> - <span style="color:red">**실제값과 관계없이 `예측값들의 퍼진 정도`만 의미!**</span>
>

## 데이터에서 편향-분산 상충관계

> **"`비용함수 = 편향 + 분산`를 줄이는 방향으로 `알고리즘이 진화하고 성능이 향상되는 편(이론에서 확인예정)`이지만, <br>
알고리즘 성능은 `데이터에 의존`되어 있기 때문에 고성능 알고리즘이 `무조건 나의 데이터에서 성능이 좋을거라 단정할 순 없고`, <br>
나의 데이터에 `여러 알고리즘을 적용하여 성능 분포나 변화를 확인`하여, <br> `과소/과대적합 되지 않는` 또는 편향과 분산이 모두 최소가 되는 `최적 알고리즘을 선택`하는것이 최적 모델링 방향"**

---

- **평균-분산 상충관계(Bias-variance Trade-off):** `복잡도가 증가`할수록 `Bias는 감소`하나 `Variance가 증가`하는 반비례 관계

> - **Train:** `복잡도가 증가`할수록 `Bias 감소 + Variance 감소`
> - **Test:** `복잡도가 증가`할수록 `Bias 감소 가능 + Variance 증가`
>
> - **Underfitting:** Train 패턴 `적게 학습`하여, `주로 Bias 때문에 Train/Test 성능 낮음`
> - **Overfitting:** Train 패턴 `과하게 학습`하여, `주로 Variance 때문에 Test 성능 낮음`
>
> <center><img src='Image/Expert/Bias_Variance4.png' width='400'></center>
>
> **"알고리즘이 복잡해지면 `Bias와 Variance는 모두 감소하는 경항`이지만, `어느 시점(Train 패턴과 다른 패턴이 Test에 나타나는 시점)`부터는 Variance가 증가하여 `Test의 비용함수/에러가 증가`"**

**2) 실제 데이터분석 접근 방법:** `편향과 분산 모두 최소화`하기 위해 `반복적`으로 업데이트

> **"`Train` 데이터의 `Bias가 적절(낮게)`한지 확인 후, `Test 데이터`에 적용하여 `Variance가 적절(낮게)하도록 반복적 업데이트`"**
>
> - Train의 Bias가 높다면,  `알고리즘 복잡하게` 또는 `빅데이터(Row & Column)` 또는 `최적화`를 통해 해결
>
> - Test의 Variance가 높다면, `알고리즘 덜 복잡하게` 또는 `빅데이터(Row) & 스몰데이터(Column)` 또는 `최적화`를 통해 해결
>
> <center><img src='Image/Expert/Bias_Variance_Reduce.png' width='600'></center>
>
> - **딥러닝(인공지능 알고리즘):** 딥러닝은 `엄청나게 복잡한 모델`이며 `Bias-variance Trade-off를 피할 수 없음`
>
> - `스몰데이터`의 딥러닝은 `과대적합`되어 `High Variance`가 우려되기에, `딥러닝`으로 성능을 내기 위해선 `빅데이터`가 반드시 필요!

- **불순도 측정 지표:** `엔트로피(Entropy)` & `지니(Gini Index)`

> **(1) 엔트로피(Entropy):** 분포의 순수도(Purity) 나타내는 척도로 순도가 낮으면 `엔트로피 높고` 순도가 높으면 `엔트로피 낮음`, `무질서도`라고도 함
>
> <center><img src='Image/Expert/Ensemble_EntropyExample.png' width='600'>(https://tyami.github.io/machine%20learning/decision-tree-1-concept/)</center>
>
> <br>
>
> - **확률 $\rightarrow$ 정보량 $\rightarrow$ 엔트로피 $\rightarrow$ 정보획득** 
>
>
> - **확률(Probability):** `예상하지 못한 사건` 발생 확률
>
> - **정보량(Information):** `잘 일어나지 않는 사건`은 자주 발생하는 사건보다 `정보량 많음` 가정
>
> $$
\begin{aligned}
\text{예상못한 사건 발생 확률(Probability)} &= p(x) \\
\text{예상못한 사건으로 알게될 정보(Information)} &= I(x) = \log_2 \frac{1}{p(x)} \\
\end{aligned}
$$
>
> <center><img src='Image/Expert/Ensemble_Information.png' width='600'>(https://tyami.github.io/machine%20learning/decision-tree-2-ID3/)</center>
>
> <br>
>
> - **엔트로피(Entropy):** `어떤 사건`이 나올지 `예측하기 어려운 정도`
>
> $$
\begin{aligned}
\text{예상못한 사건으로 알게될 정보(Information)} &= I(x) = \log_2 \frac{1}{p(x)} \\
\text{예상하지 못한 평균 정보(Entropy)} &= H(S) \\
&= \sum_{i=1}^N p_i\log_2 \frac{1}{p_i}=-\sum_{i=1}^N p_i\log_2 p_i
\end{aligned}
$$
>
> - **정보획득(Information Gain):** `분류 전후`의 `변화된 정보량`을 통해 `분류성능`을 확인하며, 분류가 잘 되지 않아 별 차이가 없으면 `낮은 수치`, 잘 분류되어 변화가 많으면 `높은 수치`
>
> $$
\begin{aligned}
\text{Information Gain} &= \text{Base Entropy} - \text{New Entropy} \\
&= \text{H(S)} - \text{H(S | Each Case)} \\
\end{aligned}
$$
>
> **(2) 지니(Gini Index):** 불확실성(묶여있는정도)을 수치화 한 것으로, 여러 레이블들이 묶이면 `수치가 높고` 유사 레이블들끼리 묶이면 `수치가 낮음`
> - `경제 불평등 지수`로 처음 등장을 하였으며, `소수에게 집중되면` 수치가 높고 `개개인에게 모두 분배`되면 수치가 낮음

## **Boosting:** Bias 줄이기 위한 목적

**"여러개의 알고리즘이 `순차적으로 학습`하되, `첫 알고리즘`이 `전체 데이터`를 학습하고 앞선 알고리즘에서 잘 학습되지 않은 `오차/에러`에 중요도를 높여 `가중치`를 부여하고 반복적으로 `모델링 업데이트`를 위한 학습 및 예측"**

- **장점:** 

> - Bagging과 달리 `맞추기 어려운 문제`를 `학습`하는데 특화
> - 학습되지 않은 패턴을 계속 반영하기 `Bias를 줄이는데 효과적`
> - `Outlier`나 `Anomaly Detection`에 강함

- **단점:** 

> - `잘 학습된 특징`들은 중요도를 낮춰 `덜 고려함`
> - 학습하지 못한 데이터 패턴들을 학습하기 때문에 `직렬처리로 느린 편`

>
> **(3)** Random Forest는 Tree 들을 병렬로(Parallel) 한번에 생성하고 Tree 들 간 서로 독립적이지만, AdaBoost는 `기존 Stump가 다음 Stump에 영향`을 주며 `순서대로(Sequential) 생성`

> **(1) First Stump:**
>
> - 몸무게 기준의 `Gini Index`가 가장 작기 때문에 `몸무게 Stump`를 첫번때 Stump로 지정
> - 몸무게 오름 차순 $\rightarrow$ 인접한 몸무게 평균 $\rightarrow$ 각 평균마다 불순도 계산 $\rightarrow$ 가장 작은 Gini Index 갖는 몸무게 평균값이 `176`
>
> <center><img src='Image/Expert/AdaBoost_ExFirst.png' width='600'></center>

> **(2) Amount of Say:** 최종 `분류 모델 예측값`에 해당 Stump가 `얼마만큼의 영향`을 주는지 `가중치`로 사용
>
> <br>
> <center><img src='Image/Expert/AdaBoost_ExFirstError.png' width='600'></center>
>
> $$
\begin{aligned} 
\text{Amount of say} &=\alpha_t=\frac{1}{2}log(\frac{1-\epsilon_t}{\epsilon_t}) \\
& \text{where } \epsilon_t \text{ is the total sum of misclassified sample weight}
\end{aligned} 
$$
>
> - `X축`은 Total Error, `Y축`은 Amount of Say
> - `Total Error가 0`이면 Amount of Say는 `큰 양수`이고, `Total Error가 1`이면 Amount of Say는 `큰 음수`
> - `Total Error가 0`이면 `올바른 분류`를 한 알고리즘이고, `1`이면 `반대로 분류`를 한 알고리즘
> - `Total Error가 0.5`일 때는 `Amount of Say = 0`, 동전을 던지는 것처럼 의미가 없다는 뜻
>
> <center><img src='Image/Expert/AdaBoost_ExAS.png' width='200'></center>
>
>  $$
\begin{aligned} 
\text{Amount of say} &=\alpha_t=\frac{1}{2}log(\frac{1- 1/8}{1/8}) \\
&= 0.97
\end{aligned} 
$$

>
> - 하나의 Stump가 `잘못 분류한 샘플`에 대해 다음 Stump로 넘겨줄 때 `가중치를 높여 전달`

>

> **(4) Next Stump:** 
>
> **(4-1):** Resampling
> - 최종 Sample Weight를 기준으로 `index`를 `0~1` 사이의 구간으로 구분
> - `원데이터의 샘플 수`만큼 `0부터 1`사이의 숫자를 `Random Sampling` 하고, 이 값에 따라 선택된 `index 샘플`로 원데이터를 새롭게 구성
> - `Sample Weight가 컷던 샘플`은 더 넓은 구간값이기 때문에, `중복되어 더 자주 추출`
> - `Sample Weight`는 모든 샘플이 동일하게 `초기화`
>> - Sample Weight는 동일하지만, 잘못 분류되었던 샘플 비율이 더 높기 때문에 결과적으로 오분류 샘플의 가중치가 다른 것보다 높은 효과
>
> <br>
> <center><img src='Image/Expert/AdaBoost_ExNextStump.png' width='600'></center>
>
> **(4-2):** `(1)`적용 및 재추정하여 `Next Stump` 생성
>
> <center><img src='Image/Expert/AdaBoost_ExFirst.png' width='600'></center>

> **(5) Repetition:** 정해진 반복만큼의 Stump 생성시 까지 `(1) + (2) + (3) + (4)`를 반복하여 `모든 Stump 생성`

> **(6) Final Decision:** 정해진 반복만큼의 Stump 생성시 까지 `(1) + (2) + (3) + (4)`를 반복하여 `모든 Stump 생성`
>
> - `각 Stump`마다의 `Amount of Say` 추정
> - 최종적으로 계산된 `Amount of Say` 평균값으로 `Heart Disease`가 있다고 분류 예측
> - 모든 Stump의 예측값 $h_t(x)$에 Amount of Say $\alpha_t$를 가중치로 곱하는 회귀함수 방식인 `Hard Voting`으로 최종 예측값 선택
>>
>> $$
F(x)=\sum_{t=1}^N \alpha_t h_t(x)
$$

### Gradient Boosting Machine(GBM) 알고리즘

> **"AdaBoost 알고리즘 대비 `기능적인 변경 및 추가`"**

---

**1) 개별 알고리즘(Weak Learner) 변경:**

- AdaBoost에선 Stump를 사용했지만, GBM에선 `Restricted Tree` 사용
- `First Weak Learner`는 모든 데이터의 평균값을 갖는 하나의 초기 추정값인 `Single Leaf`에서 출발
- `나머지 Restricted Tree`는 `최대 잎의 수`를 제한을 걸어 과적합 제한

<center><img src='Image/Expert/AdaBoost_GBM.png' width='700'></center> 

---

**2) 가중치 방식 변경:** AdaBoost는 각 Weak Learner의 가중치(크기)가 다른 반면, `GBM`에선 `동일 가중치`

> **"AdaBoost Equation"**
>
> $$
F(x)=\sum_{t=1}^N \alpha_t h_t(x)
$$
>
> **"GBM Equation"**
>
> - AdaBoost는 Amount of Say를 가중치로 사용한 반면, `GBM`에선 모두 동일한 `Learning Rate` $\eta$ 사용
>
> $$
F(x)=F_0(x) + \eta \sum_{t=1}^N h_t(x)
$$
>
> <center><img src='Image/Expert/GBM_Predicted.png' width='700'></center> 


---

**3) 예측값 변경:** Tree 예측값인 $h_t(x)$가 다르기 때문

> - AdaBoost Stump들은 모두 실제 회귀/분류 값을 예측하기 때문에, 이 값들을 가중 평균하여 실제값 예측
> - `GBM`은 각 `Restricted Tree들의 예측값`은 실제 회귀/분류 값이 아닌 `이전 Tree의 오차/잔차(Pseudo Residual)`
> - `최종 예측값`은 각 모델의 `오차를 스케일링 및 합산`하여 실제값 예측값

> **(1) First Leaf:** 모든 데이터의 `평균값 대신 확률`로 초기 추정값 `Single Leaf` 생성
>

$$

### eXtreme Gradient Boosting(XGBoost) 알고리즘 

> **"GBM을 발전시킨 알고리즘이 XGBoost, Light GBM, CatBoost"**

- `XGBoost(2014년), LightGBM(2016년), CatBoost(2017년)`은 더 빠르고 성능 좋은 알고리즘으로, Kaggle에서 상위권 Ranker들이 많이 사용하는 알고리즘
- GBM의 `Overfitting 문제를 보완`하기 위해 개발
- `분류` 뿐만 아니라 `회귀`에서도 뛰어난 `예측 성능`
- `CPU, GPU, 병렬처리` 등을 지원하여 GBM 대비 `빠른 실행시간`
- `Regularization, Tree Pruninng, Cross-validation` 등 다양항 성능 향상 최적화 기능
- `max_depth`로 분할 깊이를 조정하지만, `Tree Pruning`으로 더 이상 이득 없는 분할 수를 더 줄이는 장점

> - Approximate Greedy Algorithm
> - Parallel learning
> - Weighted Quantile Sketch
> - Sparsity-Aware Split Finding
> - Cache-Aware Access
> - Blocks for Out-of-Core Computation

- Regularization을 포함한 `다양한 비용함수` 지원
- `Early Stopping, 자체 교차 검증, 결측치 처리` 등 다양한 편의 기능
- `튜닝`해야 할 `하이퍼파라미터가 너무 많아` 단점

> - Pseudo Residual 추정을 위해 `실제값(True:1, False:0)`과 `초기 추정값 확률 차이`로 계산
>
> <center><img src='Image/Expert/XGBoost_ClassFirstResidual.png' width='500'></center> 
>
> **(2) Next Leaf:** 
>
> - `잔차 품은 유사성 이득(Gain)`기반 Decision Tree 생성
> - 부모 노드와 자식 노드의 유사성 정도 차이 변화량으로 `Gain` 추정 및 분류
>
> <center><img src='Image/Expert/XGBoost_ClassNextTree.png' width='700'></center> 
> <center><img src='Image/Expert/XGBoost_ClassNextTreeSet.png' width='700'></center> 
>
> - 과적합 대응을 위한 `Gain 기준` 하이퍼 파라미터 $\gamma$ 적용하여 `기준보다 낮은 마디`에서는 `가지치기`하여 분기를 진행하지 않음
>
> - 과적합 대응을 위해 분모에 있는 $\lambda$를 높이면, `유사성 및 분류 Gain`을 낮추어 `가지치기`를 유도하는 `하이퍼파리미터`
>
> - Decision Tree 기반 예측
> - `log(odds)`로 간주할 수 있는 `출력값`을 받아 변환하여 `예측 확률값` 획득
>
> $$
\begin{aligned}
\text{Output Value}=\frac{\sum \text{Residual}_i}{\sum[\text{Previous Probability}_i \times (1-\text{Previous Probability}_i)]+\lambda}
\end{aligned}
$$
>
> <center><img src='Image/Expert/XGBoost_ClassPredOdds.png' width='700'></center> 
> <center><img src='Image/Expert/XGBoost_ClassPredProb.png' width='700'></center> 
>
> **(3) Repeat Leaf Extension:** `확률 잔차`에 `(2)` 과정을 `순차적으로 반복`하여 `예측 업데이트`
>
> $$
F(x)=F_0(x) + \eta \sum_{t=1}^N h_t(x)
$$
>
> - `최대 반복 횟수` 또는 `확률 잔차가 더이상 작아지지 않을때`까지 반복 후 `최종 예측값` 선정
> - 모든 Tree의 `log(odds)`를 합한 후 `확률로 변환`하여 최종 예측값 획득
>

### LightGBM 알고리즘

- GBM은 모든 데이터를 탐색/학습하기에 `학습 시간이 오래걸리는 문제`를 보완하기 위해 개발
- XGBoost의 장점은 계승하고 `단점은 보완`하는 방식으로 개발
- `병렬 컴퓨팅`을 지원하며 최근에는 CPU외에 `GPU`까지 사용 가능
- 메인 기술은 `GOSS(Gradient-based One-Side Sampling)`로 Information Gain을 가중치가 작은 부분의 데이터에 집중하기 때문에 `학습이 덜 된 데이터`에 초점
- 현존하는 부스팅 알고리즘 중 `가장 빠르고 높은 예측력` 제공(특히 속도)
- 스몰데이터(약10000건 정도)에서 `과적합` 우려 
- GBM은 항상 균형을 이루는 Binary 분할을 하는 `Level-wise(균형 트리 분할)` 방식을 사용하나, LightGBM은 `Leaf-wise(리프 중심 트리 분할)`을 사용하기 때문에 시간과 메모리 모두 절약 가능

|  | **Bagging** | **Boosting** |
|:---:|:---:|:---:|
| **설명** | Bootstrapping을 통해 여러 학습 데이터를 만들고,<br> 각 학습데이터의 예측치의 평균으로 최종 예측값 선정 | 여러 알고리즘들에 데이터를 순차적으로 학습 및<br> 오차를 줄이며 최종 예측값 선정 |
| **목적** | Bias가 낮은 알고리즘들을 사용하여<br> `예측 Variance를 줄이는 것` | Variance가 낮은 알고리즘들을 사용하여<br> `예측 Bias를 줄이는 것` |
| **특징** | `병렬` 앙상블 모델<br>(각 모델은 서로 독립) | `연속` 앙상블 모델<br>(이전 모델의 오류 반영) |
| **적합한 상황** | Low Bias + `High Variance` | `High Bias` + Low Variance |

---

> **"경험적으로 `Boosting`이 Bagging보다 성능이 좋은 경우가 많지만, 엄청난 컴퓨팅 학습시간과 파라미터 튜닝에 소요되는 시간과 비용을 제외한다면... `Bagging`이 가성비라고 볼 수도 있음"**
>
> - 책상에서 `테스팅`은 Bagging
> - 실제 비즈니스 현장에 `릴리즈`는 Boosting

## **Stacking:** Bagging 또는 Boosting과 유사

> **"`개별 알고리즘`을 서로 `결합하여 예측`"**

---

- 배깅 또는 부스팅과 유사하지만, 가장 큰 차이는 `개별 알고리즘`으로 `예측한 데이터`를 학습하여 `다시 예측`
- 개별 알고리즘의 예측 데이터를 `메타 데이터`라고 지칭하는 편
- 메타 데이터를 `별도의 알고리즘`으로 `최종 학습` 후 `최종 예측`하며 이때의 알고리즘을 `메타 알고리즘`라고 지칭하는 편
- 스태킹 모델링 시, `2종류 알고리즘` 필요

> (1) 개별적인 모델링에 사용할 `개별 알고리즘`
>
> (2) 개별 기반 예측 데이터를 학습하고 예측하는데 필요한 `메타 알고리즘`

<center><img src='Image/Expert/Stacking_Structure.png' width='500'>(http://rasbt.github.io/mlxtend/user_guide/classifier/StackingClassifier/)</center> 

- `캐글과 같은 대회`에서 높은 순위를 차지하기 위해 성능 수치를 높여야 할 경우 자주 사용됨
- 일반적으로 `성능이 비슷한 모델을 결합`해서 좀더 나은 `성능 향상`을 도출
- 2~3개의 개별 모델만을 결합해서 쉽게 예측 성능을 향상시킬 수 없음
- 스태킹을 적용한다고 해서 반드시 성능 향상 된다는 보장 없음
- `과적합` 단점을 개선하기 위해 다양한 `CV(Cross Validation)` 방식을 사용한, 다양한 `CV기반 Stacking` 알고리즘 존재

**1) SHAP(SHapley Additive exPlanations) 알고리즘 배경:** 설명 가능한 AI를 위한 방법론

- **AI 블랙박스 이슈:** 많은 AI 모델(특히 딥러닝, 랜덤 포레스트, XGBoost 등)은 예측 성능이 뛰어나지만, 내부 동작을 이해하거나 해석하기 어려운 `블랙박스 모델로 간주`

<center><img src='Image/Advanced/Blackbox_vs_Whitebox.webp' width='600'>(https://liquidity-provider.com/articles/demystifying-black-box-ai-and-its-use-cases/)</center>

> - 모델들이 어떻게 결정을 내리는지 이해하지 못하면, `AI의 신뢰성이나 공정성을 평가하기 어려움`
> - 특히 `의료, 금융, 법률 등 중요한 분야일수록 그 문제가 더욱 중요`하며 해결 필요성이 높음
> - SHAP은 `게임 이론에서 유래한 Shapley 값을 기반`으로 `각 변수가 모델의 예측에 미친 영향을 정량적으로 설명`
> - SHAP는 `2017년에 Scott Lundberg & Su-In Lee에 의해 소개`

---

**2) 필요성:**

<center><img src='Image/Advanced/Blackbox_vs_Whitebox_Example.webp' width='600'>(https://www.analyticsvidhya.com/blog/2019/08/decoding-black-box-step-by-step-guide-interpretable-machine-learning-models-python/)</center>

**(1) 모델 해석 가능성 제공:** 예측을 할 때 `각 변수(Feature)가 얼마나 영향력이 큰지를 시각화`하여 결정 과정을 명확하게 설명

**(2) 모델 공정성 검토:** 특정 특성(예: 성별, 나이 등)이 얼마나 예측에 영향을 미쳤는지 분석 함으로써 `모델의 편향과 공정성 여부를 파악`

**(3) 신뢰성 향상:** 사용자는 모델의 예측이 `왜 그런 결과를 도출했는지 이해할 수 있기에 신뢰도가 높아짐`

---

**3) 특징 및 세부기능:**

- **특징**

> - Shapley 값은 게임 이론에서 유래된 개념으로, `각 참여자가 게임에서 얻은 총 이익을 공정하게 분배하는 방법`
> - AI에서 Shapley 값은 `각 변수가 모델 예측에 기여하는 정도`를 정의하는 데 사용
> - Shapley 값은 `모든 가능한 변수 조합에 대해 각 변수가 예측에 미친 영향을 계산`
> - **Local(Individual) Explanation:** 변수 뿐만 아니라 `각 고객별(Sample) 변수들의 기여도로 개별 예측 설명` 유용
> - **Global(Total) Explanation:** 고객별 개별 예측 설명을 합산하여 `전체 데이터셋에서 변수들의 모델 예측 영향을 설명`

- **세부기능:**

> **(1) Decison & Force Plot:** `개별 예측`을 이끌어낸 `변수들의 SHAP 기여도를 직관적 시각화`

**4) 요약:**

- SHAP는 `AI 모델의 예측을 투명하게 설명하는 현존하는 강력한 도구`
- SHAP 값을 통해 모델의 `예측을 직관적으로 이해하고, 모델의 공정성, 신뢰성, 투명성을 개선`
- SHAP는 설명 가능한 AI의 핵심 기술로 자리잡고 있으며, `다양한 분야에서 실용적으로 활용`

## 회귀 문제

### `Data Preprocessing`

````python
# 예제 데이터 로딩
rng = np.random.RandomState(123)
X_train = np.sort(5 * rng.rand(500, 1), axis=0)
Y_train = np.sin(X_train).ravel()                        # sin함수의 예측을 목표로한다
X_test = np.arange(0.0, 5.0, 5/X_train.shape[0])[:, np.newaxis]
Y_test = Y_train.copy()
Y_test[::5] = Y_test[::5] + 3 * (0.5 - rng.rand(100))

# 예제 데이터 시각화
plt.figure(figsize=(10,10))
plt.scatter(X_test, Y_test, s=20, c='darkorange')
plt.title('Example Data')
plt.xlabel('X')
plt.ylabel('Y')
plt.axis("equal")
plt.grid()
plt.show()
````

### `Voting` 알고리즘

````python
# Weak Learners
model_reg = LinearRegression()
model_knn = KNeighborsRegressor()

# 모델링: Voting
## 학습
res = [('LR', model_reg), ('KNN', model_knn)]
model_voting = VotingRegressor(estimators=res)
model_voting.fit(X_train, Y_train)
## 예측
Y_trpred = model_voting.predict(X_train)
Y_tepred = model_voting.predict(X_test)
## 학습결과
plt.figure(figsize=(10,10))
plt.scatter(X_test, Y_test, s=20, c='darkorange')
plt.plot(X_test, Y_tepred, color='yellowgreen', linewidth=5, label='Predict')
plt.title('Voting')
plt.xlabel('X')
plt.ylabel('Y')
plt.axis("equal")
plt.grid()
plt.legend()
plt.show()

# 성능검증
Score_voting = evaluation_reg_trte(Y_train, Y_trpred, Y_test, Y_tepred)
Score_voting
````

### `Decision Tree` 알고리즘

````python
# 모델링: Decision Tree
## 학습
model_dt = DecisionTreeRegressor(criterion='squared_error', max_depth=2,
                                 random_state=123)
model_dt.fit(X_train,Y_train)
## 학습트리 시각화 및 출력
plt.figure(figsize=(15,15))
plot_tree(model_dt, impurity=True, filled=True, rounded=True)
plt.show()
print(export_text(model_dt))
## 예측
Y_trpred = model_dt.predict(X_train)
Y_tepred = model_dt.predict(X_test)
## 학습결과
plt.figure(figsize=(10,10))
plt.scatter(X_test, Y_test, s=20, c='darkorange')
plt.plot(X_test, Y_tepred, color='yellowgreen', linewidth=5, label='Predict')
plt.title('Decision Tree')
plt.xlabel('X')
plt.ylabel('Y')
plt.axis("equal")
plt.grid()
plt.legend()
plt.show()

# 성능검증
Score_dt = evaluation_reg_trte(Y_train, Y_trpred, Y_test, Y_tepred)
Score_dt
````

````python
# 모델링: Bagging with Hyperparameter Tuning
# Hyper parameter가 너무 많아 수동으로 하나하나 시도해보며 찾아주기가 어려움
# GridSearchCV 함수를 사용하여 최적 하이퍼 파리미터를 찾고 이를 사용해 모델링
model_rf = RandomForestRegressor(random_state=123)
params = {'n_estimators': [100],    #100개의 sub Dataset, 결정 트리의 갯수로 증가시 시간 증가
          'max_depth': [5, 7, 9, 11],    # 트리의 최대 깊이로 깊어질수록 과적합 우려
          'min_samples_leaf': [5, 7, 9, 11],    # 하위 노드 분할시 최소 샘플 수
          'min_samples_split': [5, 7, 9, 11]}    # 리프 노드가 되기 위한 최소 샘플 수
model_rf_cv = GridSearchCV(estimator=model_rf, param_grid=params, 
                           cv=5,    # CV 반복 횟수
                           n_jobs=-1)    # n_jobs=-1이면 모든 cpu사용 학습
model_rf_cv.fit(X_train, Y_train)
print("최적 하이퍼 파라미터: ", model_rf_cv.best_params_)

# 모델링: Bagging
## 학습
model_rf_cv_final = RandomForestRegressor(**model_rf_cv.best_params_)
model_rf_cv_final.fit(X_train, Y_train)
## 예측
Y_trpred = model_rf_cv_final.predict(X_train)
Y_tepred = model_rf_cv_final.predict(X_test)
## 학습결과
plt.figure(figsize=(10,10))
plt.scatter(X_test, Y_test, s=20, c='darkorange')
plt.plot(X_test, Y_tepred, color='yellowgreen', linewidth=5, label='Predict')
plt.title('Random Forest with Cross Validation')
plt.xlabel('X')
plt.ylabel('Y')
plt.axis("equal")
plt.grid()
plt.legend()
plt.show()

# 성능검증
Score_rfcv = evaluation_reg_trte(Y_train, Y_trpred, Y_test, Y_tepred)
Score_rfcv
````

````python
# 구조적 이해
# Weak Learners
model_knn = KNeighborsClassifier()
model_svm = SVC()
model_rf = RandomForestClassifier()
model_xgb = XGBClassifier(eval_metric='mlogloss')

clf = [model_knn, model_svm, model_rf, model_xgb]
Y_trpred_clf, Y_tepred_clf = [], []
for algo in clf:
    algo.fit(X_train, Y_train)
    Y_trpred = algo.predict(X_train)
    Y_trpred_clf.append(Y_trpred)
    Y_tepred = algo.predict(X_test)
    Y_tepred_clf.append(Y_tepred)
    print('Test 예측 성능: {0:.4f}'.format(accuracy_score(Y_test, Y_tepred)))
X_train_stack = np.array(Y_trpred_clf).T
X_test_stack = np.array(Y_tepred_clf).T

# 모델링: Meta Learner
model_final = LogisticRegression()
model_final.fit(X_train_stack, Y_train)    # 메타 Train
Y_trpred = model_final.predict(X_train_stack)
Y_tepred = model_final.predict(X_test_stack)

# 성능검증
Score_stack = evaluation_class_AccuracyAUC(Y_train, Y_trpred, Y_test, Y_tepred)
Score_stack
````
